## `PercentSalaryHike`

```python
df['PercentSalaryHike'].describe()
```

```python
sns.histplot(df['PercentSalaryHike'], kde=True)
plt.title('Distribution of Percent Salary Hike')
plt.xlabel('Percent Salary Hike')
plt.ylabel('Frequency')
plt.show()
```


- The histogram shows that the distribution of salary hikes is slightly right-skewed, with most employees receiving hikes in the `12%` to `20%` range.
- `Mode`: The most common salary hike percentage is around 15%.
- `Outliers`: There are a few employees who received higher salary hikes (above 20%), but these are relatively rare.
- The majority of employees receive moderate salary hikes, with a small number of employees receiving significantly higher hikes.

```python
sns.boxplot(x=df['PercentSalaryHike'])
plt.title('Boxplot of Percent Salary Hike')
plt.xlabel('Percent Salary Hike')
plt.show()
```


- `Median`: The median salary hike is around `15%`.
-  The middle `50%` of employees received salary hikes between `14%` `&` `16%`.
- The salary hike distribution is relatively consistent, with most employees receiving hikes close to the median.

```python
sns.boxplot(x='JobRole', y='PercentSalaryHike', data=df)
plt.title('Percent Salary Hike by Job Role')
plt.xlabel('Job Role')
plt.ylabel('Percent Salary Hike')
plt.xticks(rotation=90)
plt.show()
```


- `Highest Hikes`: Employees in Managing Director and Research Director roles tend to receive the highest salary hikes (around `20%` or more).

- `Lowest Hikes`: Employees in Sales Representative and Human Resources roles tend to receive the lowest salary hikes (around `12%` to `14%`).

- Salary hikes vary significantly by job role, with leadership and specialized roles receiving higher hikes compared to entry-level or support roles.

```python
sns.boxplot(x='Attrition', y='PercentSalaryHike', data=df)
plt.title('Percent Salary Hike by Attrition')
plt.xlabel('Attrition')
plt.ylabel('Percent Salary Hike')
plt.show()
```


- Employees Who Left `Attrition = Yes`: These employees received slightly lower median salary hikes around `14%` compared to those who stayed `Attrition = No`, who received hikes around `15%`..
- Employees who leave the company tend to have received slightly lower salary hikes, suggesting that salary hikes may play a role in employee retention.
